# Tarea 1: Carga de Datos

```python
import pandas as pd
df = pd.read_csv("/content/Automobile.csv")
df.head()

```

# Analisis de Datos

```python
print(df.info())
print(df.describe())
```


# Tarea 2: Grafico de Dispersion

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(8, 5))
plt.scatter(df['weight'], df['mpg'], alpha=0.7, c='crimson', edgecolors='k')
plt.title('Relación entre Consumo (mpg) y Peso del Automóvil (weight)', fontsize=14)
plt.xlabel('Peso (weight)', fontsize=12)
plt.ylabel('Millas por galón (mpg)', fontsize=12)
plt.show()
```

# Tarea 3: Histograma de Consumo


```python
import matplotlib.pyplot as plt
plt.figure(figsize=(8, 5))
plt.hist(df['mpg'].dropna(), bins=15, color='skyblue', edgecolor='black', alpha=0.8)
plt.title('Distribución del Consumo de Combustible (mpg)', fontsize=14)
plt.xlabel('Millas por galón (mpg)', fontsize=12)
plt.ylabel('Frecuencia', fontsize=12)
plt.show()
```

# Tarea 4:Diagrama de Caja

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
sns.boxplot(data=df, x='cylinders', y='mpg', hue='cylinders', palette='Set2', legend=False)
plt.title('Valores atípicos: MPG por Cilindros')
plt.xlabel('Cilindros (cylinders)')
plt.ylabel('Millas por galón (mpg)')

plt.subplot(1, 2, 2)
sns.boxplot(data=df, x='origin', y='mpg', hue='origin', palette='Set3', legend=False)
plt.title('Valores atípicos: MPG por Origen')
plt.xlabel('Origen (origin)')
plt.ylabel('Millas por galón (mpg)')

plt.tight_layout()
plt.show()
```

# Tarea 5: Gráfico de Barras

```python
import pandas as pd
import matplotlib.pyplot as plt
tabla = pd.crosstab(df['model_year'], df['origin'])
tabla.plot(kind='bar', figsize=(10, 5))
plt.title('Cantidad de automóviles por año de modelo y origen')
plt.xlabel('Año del modelo (model_year)')
plt.ylabel('Cantidad')
plt.legend(title='Origen', labels=['EE.UU.', 'Europa', 'Asia'])

plt.show()
```
